```python
import pandas as pd
import numpy as np
```

```python
questions_raw = pd.read_csv("train_data.csv")
answers_raw = pd.read_csv("train_answers.csv")
```

```python
print(questions_raw.head())
```

```python
print(answers_raw.head())
```

```python
df = questions_raw.merge(answers_raw, on="id")

print(df.head())
```

```python
rows = []

for _, row in df.iterrows():
    statement = row["FalseSent"]
    correct = row["answer"]

    for i in  ["A","B","C"]:
        explanation = row[f"Option{i}"]

        label = 1 if i == correct else 0

        rows.append({
            "statement": statement,
            "explanation": explanation,
            "label": label
        })

binary_df = pd.DataFrame(rows)

print(binary_df.head())
```

```python
binary_df["input_text"] = (
    "statement: " + binary_df["statement"] + 
    " explaination"
)
```
